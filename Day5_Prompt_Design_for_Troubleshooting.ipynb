```python
# Test the troubleshooting prompt with a sample customer issue

issue = "Slow Internet"
symptoms = "Websites and applications are loading slowly"

prompt = troubleshooting_prompt.format(
    issue=issue,
    symptoms=symptoms
)

print("Troubleshooting Prompt:")
print(prompt)
```

```python
# Test the service analysis prompt with sample customer information

issue = "Internet connection problem"
churn_indicator = "Short tenure"
service_type = "Fiber optic"
tenure = 6
monthly_charges = 85.50

prompt = service_analysis_prompt.format(
    issue=issue,
    churn_indicator=churn_indicator,
    service_type=service_type,
    tenure=tenure,
    monthly_charges=monthly_charges
)

print("Service Analysis Prompt:")
print(prompt)
```

```python
# Create a combined prompt for technician troubleshooting and service analysis

def create_technician_prompt(
    issue,
    symptoms,
    churn_indicator,
    service_type,
    tenure,
    monthly_charges,
    troubleshooting_info
):

    prompt = f"""
You are a technician support assistant.

Customer Technical Issue:
{issue}

Symptoms:
{symptoms}

Churn Indicator:
{churn_indicator}

Service Type:
{service_type}

Customer Tenure:
{tenure} months

Monthly Charges:
{monthly_charges}

Relevant Knowledge Base Information:
{troubleshooting_info}

Analyze the customer situation and provide:

1. Technical issue analysis
2. Possible cause
3. Troubleshooting steps
4. Recommended solution
5. Customer retention consideration
6. Priority level

Give a clear and concise response suitable for a technician.
"""

    return prompt


print("Combined technician prompt created successfully!")
```

```python
# Test the combined technician prompt

technician_prompt = create_technician_prompt(
    issue="Slow Internet",
    symptoms="Customer reports slow browsing and frequent buffering",
    churn_indicator="Short tenure",
    service_type="Fiber optic",
    tenure=5,
    monthly_charges=79.99,
    troubleshooting_info="Check internet speed, router status, connected devices, and signal strength."
)

print(technician_prompt)
```
