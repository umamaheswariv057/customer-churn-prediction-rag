# Build the Technician Knowledge Base

## Objective

The objective of this section is to build a structured technician knowledge
base containing common customer technical issues, symptoms, troubleshooting
steps, solutions, and priority levels.

The knowledge base can be used by the technician RAG workflow to retrieve
relevant troubleshooting information based on customer issues.

```python
# Import Pandas for data analysis and DataFrame creation

import pandas as pd

print("Pandas imported successfully!")
```

```python
technician_knowledge = [
    {
        "Issue": "Internet Connection Problem",
        "Category": "Network",
        "Symptoms": "No internet connection",
        "Troubleshooting": "Check router power, network cable, and restart the router.",
        "Solution": "Restore the router connection and verify network connectivity.",
        "Priority": "High"
    },
    {
        "Issue": "Slow Internet",
        "Category": "Network",
        "Symptoms": "Websites and applications load slowly",
        "Troubleshooting": "Check internet speed, connected devices, router status, and signal strength.",
        "Solution": "Reduce unnecessary connected devices and restart the router.",
        "Priority": "Medium"
    },
    {
        "Issue": "Wi-Fi Not Working",
        "Category": "Wi-Fi",
        "Symptoms": "Device cannot connect to Wi-Fi",
        "Troubleshooting": "Check Wi-Fi settings, restart the router, and reconnect the device.",
        "Solution": "Restore Wi-Fi connectivity and verify the device connection.",
        "Priority": "High"
    },
    {
        "Issue": "Billing Problem",
        "Category": "Billing",
        "Symptoms": "Incorrect or unexpected bill",
        "Troubleshooting": "Check billing details, payment history, and subscribed services.",
        "Solution": "Verify the billing information and correct any billing discrepancy.",
        "Priority": "Medium"
    },
    {
        "Issue": "Service Disconnected",
        "Category": "Service",
        "Symptoms": "Customer service is unavailable",
        "Troubleshooting": "Check account status, payment status, and service availability.",
        "Solution": "Identify the reason for disconnection and restore service when appropriate.",
        "Priority": "High"
    },
    {
        "Issue": "Router Problem",
        "Category": "Hardware",
        "Symptoms": "Router frequently disconnects or restarts",
        "Troubleshooting": "Check power supply, cables, router status, and restart the router.",
        "Solution": "Stabilize the router connection or escalate for hardware support.",
        "Priority": "High"
    }
]

print("Technician knowledge base created successfully!")
```

```python
# Convert the technician knowledge base into a DataFrame

knowledge_df = pd.DataFrame(technician_knowledge)

knowledge_df
```

```python
# Display basic information about the technician knowledge base

print("Number of records:", knowledge_df.shape[0])
print("Number of columns:", knowledge_df.shape[1])

print("\nKnowledge Base Columns:")
print(list(knowledge_df.columns))
```

```python
# Create a function to search the technician knowledge base

def search_knowledge_base(query):
    query = query.lower()

    results = knowledge_df[
        knowledge_df["Issue"].str.lower().str.contains(query, na=False) |
        knowledge_df["Category"].str.lower().str.contains(query, na=False) |
        knowledge_df["Symptoms"].str.lower().str.contains(query, na=False)
    ]

    return results
```

```python
# Search the knowledge base for a Wi-Fi-related issue

query = "Wi-Fi"

results = search_knowledge_base(query)

print("Search Query:", query)
print("\nRelevant Knowledge:")

display(results)
```

```python
# Search the knowledge base for an internet-related issue

query = "Internet"

results = search_knowledge_base(query)

print("Search Query:", query)
print("\nRelevant Knowledge:")

display(results)
```
